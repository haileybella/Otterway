# Otterway cleaned location metadata

Run the first cell to keep eight columns from the original metadata CSV. The export keeps one row per image, including shared locations and rows with blank coordinates. It writes the cleaned CSV to the notebook’s working directory.

If your source CSV is stored somewhere else, edit the source path in the first cell. The second cell shows a preview. The cleaned CSV can also be used directly.


```python
from pathlib import Path
import csv

source = Path.home() / "Desktop/GB10competition/image_metadata(in).csv"
output = Path("otterway_locations_cleaned.csv")

columns_to_keep = [
    "image_id",
    "image_path",
    "latitude",
    "longitude",
    "locality",
    "state",
    "country",
    "location_precision",
]

with source.open(encoding="utf-8-sig", newline="") as infile:
    reader = csv.DictReader(infile)
    missing = [col for col in columns_to_keep if col not in (reader.fieldnames or [])]

    if missing:
        raise ValueError(f"These columns weren't found in the original CSV: {missing}")

    with output.open("w", encoding="utf-8", newline="") as outfile:
        writer = csv.DictWriter(outfile, fieldnames=columns_to_keep)
        writer.writeheader()
        writer.writerows(
            {col: row[col] for col in columns_to_keep}
            for row in reader
        )

print(f"Created {output}")
```

```python
with output.open(encoding="utf-8-sig", newline="") as f:
    location_rows = list(csv.DictReader(f))

print(f"Loaded {len(location_rows):,} rows and {len(columns_to_keep)} columns")
location_rows[:5]
```
